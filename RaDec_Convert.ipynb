```python
# convert ra/dec from hms to decimal degrees

def sexagesimal_to_decimal(ra, dec):
    # Convert RA from hh:mm:ss.s to decimal degrees
    ra_h, ra_m, ra_s = map(float, ra.split(':'))
    ra_decimal = (ra_h + ra_m / 60 + ra_s / 3600) * 15

    # Convert Dec from ±dd:mm:ss.s to decimal degrees
    dec_sign = -1 if dec.startswith('-') else 1
    dec_d, dec_m, dec_s = map(float, dec.lstrip('+-').split(':'))
    dec_decimal = dec_sign * (dec_d + dec_m / 60 + dec_s / 3600)

    return ra_decimal, dec_decimal

def process_file(input_filename, output_filename):
    with open(input_filename, 'r') as infile, open(output_filename, 'w') as outfile:
        for line in infile:
            # Clean and normalize the line
            line = line.strip().rstrip('\\').strip('}').strip()
            if not line:
                continue
            try:
                ra_str, dec_str = map(str.strip, line.split(','))
                ra_dec, dec_dec = sexagesimal_to_decimal(ra_str, dec_str)
                outfile.write(f"{ra_dec:.6f},{dec_dec:.6f}\n")
            except Exception as e:
                print(f"Skipping malformed line: {line} — Error: {e}")
```
